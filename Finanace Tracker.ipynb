```python

# =====================================
# FINTRACK PRO – CLI FINANCE MANAGER
# =====================================

# ---------- IMPORTS ----------
# Use to create Database and SQL Queries
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey, text

#Use to define orm table and sessions
from sqlalchemy.orm import declarative_base, sessionmaker, relationship

#-------------DATABASE CONNECTION-----------
engine = create_engine("sqlite:///finance.db")

#Base Class
Base = declarative_base()

#Creating session
Session = sessionmaker(bind=engine)

#creating session object
session = Session()

#-------------TABLE-----------
#Tables:
#1. categories(id, name)  
#2. expenses(id, title, amount, date, category_id)  
#3. subscriptions(id, name, amount, next_date)  
#4. budgets(id, month, limit)

#Category Table
class Category(Base):
    __tablename__ = "categories"
    id = Column(Integer, primary_key=True)
    name = Column(String)
    expenses = relationship("Expense", back_populates="categories")
#    print("Category Added")

#Expense Table
class Expense(Base):
    __tablename__ = "expenses"
    id=Column(Integer, primary_key=True)
    title = Column(String)
    amount = Column(Integer)
    date = Column(String)
    category_id = Column(Integer, ForeignKey("categories.id"))
    categories = relationship("Category", back_populates="expenses")
#    subscriptions = relationship("Subscription", back_populates="expense")
#    print("Expense Added")

#Subscription Table
class Subscription(Base):
    __tablename__ = "subscriptions"
    id = Column(Integer, primary_key=True)
    name = Column(String)
    amount = Column(Integer)
    next_date = Column(String)
 #   expense_id = Column(Integer, ForeignKey("expenses.id"))
 #   expense = relationship("Expense", back_populates="subscriptions")
 #   print("Subscription Added")

#Budget Table
class Budget(Base):
    __tablename__ = "budgets"
    id = Column(Integer, primary_key=True)
    month = Column(String)
    limit = Column(Integer)

#Create Table in Database
Base.metadata.create_all(engine)

#-----------Functions-------------

def add_category():
    #Ask user for category name
    name = input("Category Name:")

    #Create category object and save
    session.add(Category(name=name))
    session.commit()
    print("category Added")

def add_expense():
    #Ask user for title and expense
    title = input("Expense Title:")
    amount = int(input("Amount:"))
    date = input("Date (DD-MM-YYY):")
    
    #Create Expense object and save
    session.add(Expense(title=title, amount=amount, date=date))
    session.commit()
    print("Expense Added")

def add_subscription():
    #Ask user for Subscription
    name = input("Subscription:")
    amount = int(input("Amount:"))
    date = input("Next Date:")

    #Create Subscription object and save
    session.add(Subscription(name=name, amount=amount, next_date=date))
    session.commit()
    print("Subscription Added")

def update_expense():
    eid = int(input("Expense Id:"))

    #expense record
    expense = session.query(Expense).filter(Expense.id == eid).first()

    if expense:
        expense.amount = int(input("Updated Amount:"))
        expense.date = input("Update Date:")
        session.commit()
        print("Expense Updated")
    else:
        print("Expense not found")

def delete_expense():
    eid = int(input("Expense ID:"))

    expense = session.query(Expense).filter(Expense.id == eid).first()

    if expense:
        session.delete(expense)
        session.commit()
        print("Expense deleted")
    else:
        print("Expense not found")

def search_by_date():
    date = input("Enter Date:")

    expenses = session.query(Expense).filter(Expense.date == date).all()

    for i in expenses:
        print(i.title,"-",i.date)

        
def monthly_budget_alert():
    month = input("Month (YYYY-MM):")
    limit = int(input("Montly limit:"))
    
    budget = session.query(Budget).filter(Budget.limit == limit).first()

    if budget > budgets.limit:
        print("monthly budget exceed")
    else:
        print("Within limit")

    
    #for i in budget:
    #    print(i.month,":",i.limit)

#--------------CLI Menu--------------

while True:
    print(""" =====FINANCE TRACKER========
    1. Add Category
    2. Add Expense
    3. Update Expense
    4. Delete Expense
    5. Search By Date
    6. Monthly Budget Alert
    7. Category Analytics
    8. Exit
    """)
    choice = input("choose: ")

    if choice == "1":
        add_category()
    elif choice == "2":
        add_expense()
    elif choice == "3":
        update_expense()
    elif choice == "4":
        delete_expense()
    elif choice == "5":
        search_by_date()
    elif choice == "6":
        monthly_budget_alert
    elif choice == "8":
        break
    else:
        print("Invalid Choice")
        
```
